```python
df_sample = pd.read_csv("nba_sample.csv")
df_sample.info()
df_sample['player_height_m'] = df_sample['player_height'] / 100
df_sample['BMI'] = (df_sample['player_weight'] / df_sample['player_height_m'] ** 2)
df_sample
```

```python
'''Create a new column with the name “BMI Evaluation”. Assign the BMI status based on the following condition,


BMI <= 18.4 - "Underweight"

BMI <= 24.9 - "Healthy"

BMI <= 29.9 - "Over Weight"

BMI <= 34.9 - "Severely Over Weight"

BMI <= 39.9 - "Obese"
39.9 - "Severely Obese."
'''

import numpy as np
df_sample['BMI_Evaluation'] = np.select([(df_sample['BMI'] <= 18.4), 
                                        ((df_sample['BMI'] > 18.4) & (df_sample['BMI'] <= 24.9)),
                                        ((df_sample['BMI'] > 24.9) & (df_sample['BMI'] <= 29.9)), ((df_sample['BMI'] > 29.9) & (df_sample['BMI'] <= 34.9)),
                                        ((df_sample['BMI'] > 34.9) & (df_sample['BMI'] <= 39.9)), (df_sample['BMI'] > 39.9)],
                                        ['Underweight', 'Healthy',
                                          'Overweight', 'Severely Overweight',
                                          'Obese', 'Severely Obese']
                                        )
df_sample.info()
print('\n')
print(df_sample['BMI_Evaluation'].value_counts())
```

```python
#Find the total salary spent by each team on their players
df.groupby(by = 'Team')['Salary'].sum().sort_values(ascending = False)
```

```python
#filter all the players with age greater than 35
#for the filtered data, find the heaviest player in the team and find the position of the player
cond = df[df['Age'] > 35]
cond.groupby(by = ['Team', 'Position'])['Weight(lbs)'].max().sort_values(ascending = False)
```

```python
import matplotlib.pyplot as plt
x = ['Boston Celtics', 'Toronto Raptors', 'Los Angeles Lakers', 'Sacramento Kings']
y = (df[(df['Team'].isin(x)) & (df['Age'] <= 26)]['Team'].value_counts())
plt.barh(x,y, color='skyblue')
plt.xlabel('Number of players')
plt.ylabel('Team Names')
plt.title('No. of players with age less than or equal to 26 from specific teams')
plt.show()
```

```python
x = ['Atlanta Hawks', 'Cleveland Cavaliers', 'New York Knicks', 'Sacramento Kings']
cond = df[df['Team'].isin(x)]
y = cond.groupby(by = 'Team')['Weight(lbs)'].max()
plt.figure(figsize = (10, 4.8))
plt.bar(x,y,color='orange')
plt.title('Player with maximum weight in specific teams')
plt.xlabel('Team Names')
plt.ylabel('Weight in lbs')
plt.show()
```

```python
x = ['Atlanta Hawks', 'Boston Celtics', 'Memphis Grizzlies', 'New Orleans Pelicans', 'New York Knicks']

cond = df[df['Team'].isin(x)]
cond.groupby(by = 'Team')['Name'].count()

y = cond.groupby(by = 'Team')['Name'].count()

plt.figure(figsize = (10,6))
plt.plot(x, y, '*-', color = 'red', linewidth = 0.5)
plt.title('Number of players in specific teams')
plt.xlabel('Team Names')
plt.ylabel('Number of players')
plt.show()
```
