```python
import numpy as np
import pandas as pd
```

```python
#Importing LogisticRegression Model
#0 - Ham email
#1 - Spam email
from sklearn.linear_model import LogisticRegression
```

```python
#Importing accuracy score function
from sklearn.metrics import accuracy_score
```

```python
#Importing train test split function
from sklearn.model_selection import train_test_split
```

```python
#To convert all the strings formats into numeric
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
#Importing the dataset
raw_mail_data = pd.read_csv('mail_data.csv')
raw_mail_data
```

```python
#Replacing all the null values with null string
mail_data = raw_mail_data.where((pd.notnull(raw_mail_data)), '')
mail_data.info()
```

```python
#Printing the 5 rows
mail_data.head()
```

```python
#Checking the rows and columns of the dataset
mail_data.shape
```

**LABEL ENCODING**

```python
#Label spam email as 0 and ham email as 1
mail_data['Category'] == 'spam'
#mail_data.iloc[0:3, 0]
#mail_data.loc[0:3, 'Category']
```

```python
mail_data.loc[mail_data['Category'] == 'spam', 'Category'] = 0
```

```python
mail_data.loc[mail_data['Category'] == 'ham', 'Category'] = 1
```

```python
mail_data
```

```python
#Find the total number of ham email and spam email
mail_data['Category'].value_counts()
```

```python
#Separating the data as texts and label
X = mail_data['Message'] #Input Feature
Y = mail_data['Category'] #Label
```

```python
print(X)
```

```python
print(Y)
```

```python
#Splitting the data into training data and testing data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 2)
print(X.shape, X_train.shape, Y_test.shape)
```

```python
#Transforming the text data to feature vectors that can be used as input to the logistic regression
feature_extraction = TfidfVectorizer(min_df = 1, stop_words = 'english', lowercase = True)
```

```python
X_train_num  = feature_extraction.fit_transform(X_train)
X_test_num = feature_extraction.transform(X_test)
```

```python
#Convert Y_train and Y_test values as integers
Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

**Training the data**

```python
model = LogisticRegression()
```

```python
#Training the Logistic Regression model with the training data
model.fit(X_train_num, Y_train)
```

```python
#Evaluating the Trained Model
X_train_prediction = model.predict(X_train_num)
#Finding the accuracy on training data
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print("Accuracy Score of training data", training_data_accuracy)
```

```python
X_test_prediction = model.predict(X_test_num)
testing_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print("Accuracy Score of testing data", testing_data_accuracy)
```

```python
#Building a predictive system
input_mail = ['Free Free Free You have won a lottery']
input_mail_num = feature_extraction.transform(input_mail)
```

```python
#Making Prediction
prediction = model.predict(input_mail_num)
if prediction[0] == 1:
  print('Ham Email')
else:
  print('Spam Email')
```
