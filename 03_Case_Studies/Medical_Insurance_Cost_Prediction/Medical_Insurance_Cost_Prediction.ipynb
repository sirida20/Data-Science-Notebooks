Data Preprocessing

Encoding the Categorical features

```python
#Encoding the sex column
pd.unique(insurance_dataset['sex'])
insurance_dataset.replace({'sex':{'male':0, 'female':1}}, inplace = True)
#Male - 0
#Female - 1

#Encoding Smoker Column
pd.unique(insurance_dataset['smoker'])
insurance_dataset.replace({'smoker':{'yes':0, 'no':1}}, inplace = True)
#Yes - 0
#No - 1

#Encoding Region Column
pd.unique(insurance_dataset['region'])
insurance_dataset.replace({'region':{'southwest':0, 'southeast':1, 'northwest':2, 'northeast':3}}, inplace = True)
#Southwest - 0
#Southeast - 1
#Northwest - 2
#Northeast - 3
```

Splitting Features and Target

```python
X = insurance_dataset.drop(columns = 'charges', axis = 1)
Y = insurance_dataset['charges']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the data into training and testing data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 2)
```

```python
print(X.shape, X_train.shape, Y_test.shape)
```

Model Training

```python
#Calling Linear Regression Model
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

Model Evaluation

```python
#Prediction on training data
training_data_prediction = model.predict(X_train)
```

```python
#R Squared Value on training data
training_accuracy =  metrics.r2_score(Y_train, training_data_prediction)
print('Accuracy Score of training data:', training_accuracy)
```

```python
#Prediction on testing data
testing_data_prediction = model.predict(X_test)
```

```python
#R squared value on testing data
testing_accuracy = metrics.r2_score(Y_test, testing_data_prediction)
print('Accuracy of testing data:', testing_accuracy)
```

Building a Predictive System

```python
input_data = (31,0,36.3,2,0,0) #38711
#Changing input_data into a numpy array
input_data_as_numpy_array = np.asarray(input_data)
#Reshaping the array
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)
```

```python
#Predictive the cost
prediction = model.predict(input_data_reshaped)
print('The Cost of Insurance is:', prediction)
```
