```python
#Importing important libraries
import numpy as np
import pandas as pd
import sklearn.datasets
```

```python
#Importing LogisticRegression Model
from sklearn.linear_model import LogisticRegression
```

```python
#Importing accuracy score function
from sklearn.metrics import accuracy_score
```

```python
#Importing dataset
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
breast_cancer_dataset
```

```python
#Load the dataset into a dataFrame
df = pd.DataFrame(breast_cancer_dataset.data, columns = breast_cancer_dataset.feature_names)
```

```python
#Print the first 5 rows
df.head()
```

```python
#Adding the Target(Label) Column to df
#Label (Supervised Learning)
df['label'] = breast_cancer_dataset.target
```

```python
#Print the last 5 rows
df.tail()
```

```python
#No. of columns and rows in the dataset
df.shape
```

```python
#Getting the info about the data
df.info()
```

```python
#Find the missing values
df.isnull().sum()
```

```python
#Statistical measures about the data
df.describe()
```

```python
#Checking the distribution of target(label) variable
#1 - Benign
#0 - Malignant
df['label'].value_counts()
```

```python
#Group by label column and find the mean for all the columns
df.groupby(by = 'label').mean()
```

```python
#Input Feature - X (All the columns except the label)
#Target Column - Y (Label Column)
#Only when dealing with Labels or Supervised Learning
```

```python
#Separating features and target columns
X = df.drop(columns = 'label', axis = 1)
Y = df['label']
```

```python
#Printing the feature
print(X)
```

```python
#Print the Target Variable
print(Y)
```

```python
#Splitting the data into training data and testing data
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 2)
print(X.shape, X_train.shape, Y_test.shape)
```

```python
#Model Training
#Logistic Regression is most useful for binary operations
model = LogisticRegression()
#Training the ML Model using training data
model.fit(X_train, Y_train)
```

```python
#Model Evaluation
#Checking Accuracy Score of training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print("Accuracy Score of training data", training_data_accuracy)
```

```python
#Checking Accuracy Score of testing data
X_test_prediction = model.predict(X_test)
testing_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print("Accuracy Score of testing data", testing_data_accuracy)
```

```python
#Building a predictive system
input_data = (13.17,18.66,85.98,534.6,0.1158,0.1231,0.1226,0.0734,0.2128,0.06777,0.2871,0.8937,1.897,24.25,0.006532,0.02336,0.02905,0.01215,0.01743,0.003643,15.67,27.95,102.8,759.4,0.1786,0.4166,0.5006,0.2088,0.39,0.1179)
input_data_as_numpy_array = np.asarray(input_data)
#Reshape the numpy array as we are predicting for one data point
input_data_reshape = input_data_as_numpy_array.reshape(1,-1) #(start, datapoint)
```

```python
#Prediction of one data point
prediction = model.predict(input_data_reshape)
if prediction == 0:
  print('Breast Cancer is Malignant')
else:
  print('Breast Cancer is Benign')
```
