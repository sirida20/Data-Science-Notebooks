Data Collection and Analysis

```python
car_dataset = pd.read_csv('car data.csv')
```

```python
#Seller Type
pd.unique(car_dataset['Seller_Type'])
car_dataset['Seller_Type'].value_counts()
```

Encoding the Categorical Data

```python
#Encoding the Fuel_Type column
car_dataset.replace({'Fuel_Type':{'Petrol':0, 'Diesel':1, 'CNG':2}}, inplace = True)
#Petrol - 0
#Diesel - 1
#CNG - 2

#Encoding the Transmission Column
car_dataset.replace({'Transmission':{'Manual':0, 'Automatic':1}}, inplace = True)
#Manual - 0
#Automatic - 1

#Encoding the Seller_Type Column
car_dataset.replace({'Seller_Type':{'Dealer':0, 'Individual':1}}, inplace = True)
#Dealer - 0
#Individual - 1
```

```python
car_dataset
```

Splitting the data into Training Data and Testing Data

```python
X = car_dataset.drop(['Car_Name', 'Selling_Price'], axis = 1)
Y = car_dataset['Selling_Price']
```

```python
#Splitting the data into training data and testing data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 2)
```

```python
print(X.shape, X_train.shape, Y_test.shape)
```

Model Training

```python
#Calling Linear Regression Model
model = LinearRegression()
```

```python
#Training the ML Model using training data
model.fit(X_train, Y_train)
```

```python
training_data_prediction = model.predict(X_train)
```

```python
'''
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print("Accuracy Score of training data", training_data_accuracy)
'''
training_data_accuracy = metrics.r2_score(Y_train, training_data_prediction)
print("Accuracy Score of training data", training_data_accuracy)
```

```python
testing_data_prediction = model.predict(X_test)
```

```python
testing_data_accuracy = metrics.r2_score(Y_test, testing_data_prediction)
print("Accuracy Score of training data", testing_data_accuracy)
```

Building a Predictive System

```python
input_data = (2010,7.98,41442,0,0,0,0)

input_data_as_numpy_array = np.array(input_data)
#Reshape the numpy array as we are predicting for one data point
input_data_reshape = input_data_as_numpy_array.reshape(1,-1)
```

```python
prediction = model.predict(input_data_reshape)
print('Car price according to the prediction:', prediction)
```
