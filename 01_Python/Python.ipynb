```python
#BMI calculator
#BMI = weight(kg) / height(m)^2
#if the BMI is less than 18.5 then underweight
#elif BMI is between 18.5 and 24.9 then print normal weight
#elif BMI is between 24.9 and 29.9 then overweight
#else obese
weight_kg = float(input("Enter weight in kg: "))
height_m = float(input("Enter height in mts: "))
BMI = weight_kg / (height_m)**2   #Squaring a number is done by exponent symbol (**)
print(BMI)
if BMI < 18.5:
  print("Underweight")
elif BMI >= 18.5 and BMI <= 24.9:
  print("Normal weight")
elif BMI > 24.9 and BMI <= 29.9:
  print("Overweight")
else:
  print("Obese")
```

```python
#Creating a game
import random
secret_number = random.randint(1,10)
#initializing variables
max_attempts = 5
attempts = 0
#start the game
print("Welcome to Number Game")
while attempts < max_attempts:
  guess = int(input("Guess the secret number: "))
  if guess == secret_number:
    print("Congratulations!")
    break
  elif guess < secret_number:
    print("Too low. Try higher number")
  elif guess > secret_number:
    print("Too high. Try lower number")
  else:
    print("You lose try again")
    attempts += 1

if attempts == max_attempts:
  print("Sorry, you've run out of attempts")
```

```python
#Multiple Plots on the same axis
#here we will see how to add 2 plots within the same axis
x = np.array([1,2,3,4,5])
y = x * 2
plt.plot(x, y, '--') #'-' is default design
x1 = [1, 2, 3, 4, 5]
y1 = [5, 4, 3, 2, 1]
plt.plot(x1, y1, '-*')
plt.xlabel('X - array')
plt.ylabel('Y - array')
plt.title('Line Plot with Multiple Plots on the same axis')
plt.show()
```

```python
#Multiple Plots on the same axis
#here we will see how to add 3 plots within the same axis
x = np.array([1,2,3,4,5])
y = x * 2
plt.plot(x, y, '-') #'-' is default design
x1 = [1, 2, 3, 4, 5]
y1 = [5, 4, 3, 2, 1]
plt.plot(x1, y1, '-*')
x2 = [1,2,3,4,5]
y2 = [6,3,9,1,8]
plt.plot(x2, y2, '--')
plt.xlabel('X - array')
plt.ylabel('Y - array')
plt.title('Line Plot with Multiple Plots on the same axis')
plt.show()
```

```python
#Fill the area between two plots
x = np.array([2,4,6,8])
y = x * 2
plt.plot(x, y, '-') #'-' is default design
x1 = [2,4,6,8]
y1 = [3,5,7,9]
plt.plot(x1, y1, '-*')
plt.xlabel('X - array')
plt.ylabel('Y - array')
plt.title('Line Plot with Multiple Plots on the same axis')
plt.fill_between(x,y,y1, color = 'green', alpha = 0.3)
#alpha changes the intensity of the color
plt.show()
```

```python
#Scatter Plot
x = [1,2,3,4,5]
y = [5,1,4,6,2]
plt.scatter(x,y)
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('Scatter Plot')
plt.show()
```

```python
#Scatter plot with different shape and color for two datasets
#Dataset - 1
x1 = [89,43,36,95,10,66,34,38,20,45]
y1 =[21,46,3,35,67,95,93,72,58,10]
#Dataset - 2
x2 = [26,29,48,64,6,5,36,66,72,10]
y2 = [26,24,90,33,38,20,56,2,47,15]

plt.scatter(x1, y1, c = 'red', linewidths = 2, edgecolors ='green', s = 50, marker = 'h')
#defalut marker is o
plt.scatter(x2, y2, c = 'yellow', linewidths = 2, edgecolors ='pink', s = 50, marker = 'D')
plt.show()
```

```python
#Bar Chart
import matplotlib.pyplot as plt
x = ['A', 'B', 'C', 'D']
y = [5,2,9,1]
plt.bar(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Graph')
plt.show()
```

```python
#Bar Chart (Horizontal)
x = ['A', 'B', 'C', 'D']
y = [5,2,9,1]
plt.barh(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Graph (Horizontal)')
plt.show()
```

```python
#Bar Graph (height)
x = ['A', 'B', 'C', 'D']
y = [5,2,9,1]
plt.barh(x,y,height=0.3)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Graph (Horizontal)')
plt.show()
```

```python
#Bar Chart (width)
import matplotlib.pyplot as plt
x = ['A', 'B', 'C', 'D']
y = [5,2,9,1]
plt.bar(x,y,width=0.3, color='red', edgecolor='green')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Graph')
plt.show()
```

```python
#Pie Chart
categories = ['A','B','C','D']
data = [10,17,11,14]
explode = [0,0,0.1,0]
plt.pie(data, labels = categories, autopct = '%1.1f%%', explode = explode, startangle=140, colors = ['yellowgreen','lightcoral','lightskyblue','lightgreen'], shadow = True)
plt.title('Pie Chart')
plt.show()
```

**Visualization with seaborn**

```python
import pandas as pd
df = pd.read_csv('nba.csv')
df
```

```python
#Visualization using seaborn
#Lineplot
import seaborn as sns
import matplotlib.pyplot as plt
sns.lineplot(x = df['Age'], y = df['Weight'], hue = df['Position'])
plt.show()
#dark blue line is mean weight
#hue helps in categorising the data
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.lineplot(x = df['Position'], y = df['Age'])
plt.show()
```

```python
#Scatter plot
sns.scatterplot(x = df['Age'], y = df['Weight'], hue = df['Position'])
plt.show()
```

```python
#Box plot
sns.boxplot(df['Age'])
#Middle Line is the mean age
```

```python
#Histogram
sns.histplot(df['Position'], kde = True)
plt.show()
```

```python
#Pair Plot
sns.pairplot(df, hue = 'Position')
plt.show()
```
