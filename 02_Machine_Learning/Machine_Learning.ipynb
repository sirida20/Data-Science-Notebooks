```python
#Step 3 (Training the Model)
#Fitting the data
Regression_model.fit(X,y)
```

```python
#Step 4 (Prediction)
predict = Regression_model.predict(X)

import matplotlib.pyplot as plt
plt.scatter(predict, y)
```

```python
#K Nearest Neighbors
from sklearn.neighbors import KNeighborsRegressor
Model_K = KNeighborsRegressor()
Model_K.fit(X,y)
predict = Model_K.predict(X)
plt.scatter(predict, y)
```

```python
df.drop(['body'], axis = 1)
```

```python
#Value imputation (Simple Imputer (Library Name))
'''
In Scikit-learn (sklearn), value imputation refers to the process of replacing missing or null values in a dataset with substituted values.
This is crucial because many machine learning algorithms cannot handle missing data and might throw errors or produce biased results if missing values are present.
'''
from sklearn.impute import SimpleImputer
print('Current Null values', df['age'].isnull().sum())
```

```python
#Method 1
imp = SimpleImputer(strategy = 'mean')
df['age'] = imp.fit_transform(df[['age']])
print('Present Null values', df['age'].isnull().sum())
```

```python
#Method 2
#1) strategy
#2) missing values
def get_parameter(df):
  parameters = {}
  for col in df.columns[df.isnull().any()]:
    if df[col].dtype == 'float64' or df[col].dtype == 'int64' or df[col].dtype == 'int32':
      strategy = 'mean'
    else:
      strategy = 'most_frequent'
    missing_values = df[col][df[col].isnull()].values[0]
    parameters[col] = {'missing_values':missing_values, 'strategy':strategy}

  return parameters

get_parameter(df)
```

```python
from sklearn.impute import SimpleImputer

parameters = get_parameter(df)

for col, param in parameters.items():
  missing_values = param['missing_values']
  strategy = param['strategy']
  imp = SimpleImputer(missing_values = missing_values, strategy = strategy)

df[col] = imp.fit_transform(df[[col]])
```
